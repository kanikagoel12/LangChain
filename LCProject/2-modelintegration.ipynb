### Streaming

Most Models can stream their output while it is being generated. By displaying output progressively, streaming significantly improves user experience, particulary for longer response. Calling stream() returns as iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real time.

```python
model.invoke("Write me a 200 words paragraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Write me a 200 words paragrpah on Artificial Intelligence"):
    print(chunk.text,end="|",flush=True)
```

```python
for chunk in model.stream("Why do parrots have colorful feathers?"):
    print(chunk.text,end="|",flush=True)
```

### Batch

Batching is a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel

```python
responses=model.batch([
    "Why do parrots have colourful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
])

for response in responses:
    print(response)
```

```python
responses=model.batch([
    "Why do parrots have colourful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
], config={
    'max_currency':5, #Limit to 5 parallel calls
})

for response in responses:
    print(response)
```
